##Creating Dataframes

```python
# using lists
student_data = [
    [100,44,50],
    [90,70,7],
    [120,100,14],
    [80,50,2]
]

print(pd.DataFrame(student_data, columns=['iq','marks','package']))
```

```python
# using dicts
student_dict = {
    'iq':[100,90,120,80,0,0],
    'marks':[44,70,100,50,0,0],
    'package':[50,7,14,2,0,0]
}

students = pd.DataFrame(student_dict)
print(students)
```
