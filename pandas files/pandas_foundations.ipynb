#Pandas All Concepts

```python
# importing pandas
import numpy as np
import pandas as pd
```

##Series made from Lists

```python
# Series -> columns in the table are called series in pandas. Series is a class inside pandas.
country = ['India','Pakistan','China','Srilanka']
pd.Series(country)

# Notes: In pandas , when we get dtype: object , it means values are in string form. Not in everycase but most of the cases.
#        Series include two things: index and values
#        country is as object of 'Series' class
```

```python
# integers
runs = [13,45,77,23,55]
pd.Series(runs)
```

```python
# custom index
marks = [67,69,99,10]
subjects = ['maths','science','english','french']

custom = pd.Series(marks,index=subjects)
print(custom)
```

```python
# setting a name
y = pd.Series(marks,index=subjects,name='Marks of Mayank')
print(y)
```

##Series made from dictionary

```python
marks = {
    'English':69,
    'Hindi':100,
    'Computer science':100
}

marks_dict = pd.Series(marks)
print(marks_dict)
```

##Series using read_csv

```python
# with 1 column
subs = pd.read_csv('/content/subs.csv')
print(subs)
```

```python
# with 2 column
vk = pd.read_csv('/content/kohli_ipl.csv',index_col='match_no')
print(vk)
```

```python
movies = pd.read_csv('/content/bollywood.csv',index_col='movie')
print(movies)
```

##Series Methods

```python
# head and tail
# head -> give data preview only , show first 5 rows from data by dafault BUT we can ask for more by mentioning no. of rows in head()

print(subs.head())
```

```python
print(vk.tail())
```

```python
# sample -> give one row from data randomly
# useful when our data has some bias
print(movies.sample(3))
```

```python
# value_counts -> tells the frequency for a value present in data in DESC order
print(movies.value_counts())
```

```python
# sort_values
print(vk.sort_values(by='runs',ascending=False))
```

```python
# we can use head() here also
# THIS IS CALLED METHOD CHAINING
print(vk.sort_values(by='runs',ascending=False).head(3).values[0])
```

```python
# if you want to do permanent changes in dataset , you have to use inplace=True
vk.sort_values(by='runs',inplace=True)
```

```python
print(vk)
```

```python
# sort_index
print(movies.sort_index())
```

##Series Maths Methods
