##BASIC FUNCTIONS

```python
# Importing numpy as np , np is just a shortform which we will use in the notebook instead of writing "numpy"
import numpy as np
```

```python
# How 1D array is made using numpy:
array_1d = np.array([1,2,3,4])
print(f"array_1d looks like: {array_1d}")
print(f"dtype of array_1d: {type(array_1d)}")
```

```python
# 2D and 3D arrays using np
array_2d = np.array([[1,2],[3,4]])
print(f"array_2d looks like:\n{array_2d}")
print(f"dtype of array_2d: {type(array_2d)}. Same as array_1d")

array_3d = np.array([[[1,2],[3,4]],[[5,6],[7,2]]])
print(f"array_3d looks like:\n{array_3d}")
print(f"dtype of array_3d: {type(array_3d)}. Same as array_1d")
```

```python
# np.arange function -> works like range function which require 2 values , 1st value is included and 2nd is exclused in output array.
print("printing the output without storing it in variable:")
np.arange(1,11)
```

```python
print("After storing it in arange_variable:")
arange_variable = np.arange(1,11)
print (f"arange_variable:\n {arange_variable}")
print("See, 1 is included in array but 11 is excluded, number before 11 is included.")
```

```python
# using dtype
np.arange(1,11,dtype=float)
```

```python
# arange with steps
np.arange(1,11,2)
```

```python
# arange with single value (it starts with 0 when positive number and [] when single negative value is given)
np.arange(10)
```

```python
np.arange(-10,10)
```

```python
# with reshape -> .reshape() helps in reshaping the numpy array with valid input. If input for .reshape is (3,3) it means reshaping array in 3x3 array or matrix.
# np.arange(1,11) include 10 values, using reshape function we can reshape the array in (2,5)...(5,2)... but we can't reshape it as (3,4)
# because 3x4 matrix include 12 values and here we have only 10.
np.arange(1,11).reshape(2,5)
```

```python
# arange with reshape example
np.arange(16).reshape(2,2,2,2)
```

```python
"""Initialising Array using np.zeros , np.ones and np.random"""

# np.ones and np.zeros
# np.ones need shape as input OR here, we provide the 'tuple'
ones_variable = np.ones((2,3))
print(f"ones_variable:\n{ones_variable}\n")

zeros_variable = np.zeros((4,5))
print(f"zeros_variable:\n{zeros_variable}\n")

print("USECASE: intializing weights in neural networks")
```

```python
# np.random -> initialize with random values between 0 and 1
np.random.random((3,4))

# why two randoms? -> because random is a class and it has a method named random ( concept of OOPS )
```

```python
# np.linspace -> helps in generating the values within a range where lowerbound and upperbound are inclusive/included.
# ALSO, all the elements generated are linearly gapped
# np.linspace(lowerbound, upperbound, number of elements)
np.linspace(-10,10,10)
```

```python
# np.identity -> helps creating identity matrix
np.identity(3)
```

##ARRAY ATTRIBUTES:
We know that every np array is a object of numpy class.
So, attributes are properties of NumPy arrays that provide information about the array's shape, size, data type, dimension, and so on

```python
# creating 3 arrays for further use
# a1 is a array/vector , a2 is a matrix , a3 is like a tensor or 3D tensor
a1 = np.arange(10,dtype=np.int32)
a2 = np.arange(12,dtype = float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
```

```python
# printing the a1,a2,a3
print(f"a1:\n{a1}\n")
print(f"a2:\n{a2}\n")
print(f"a3:\n{a3}\n")
```

```python
# ndim -> tells the dimension of array
print(f"dimension of a1: {a1.ndim}")
print(f"dimension of a2: {a2.ndim}")
print(f"dimension of a3: {a3.ndim}")
```

```python
# np.shape -> tells the shape in array

print(f"shape of a1: {a1.shape}")
print(f"shape of a2: {a2.shape}")
print(f"shape of a3: {a3.shape}")

# shape of a1: (10,) means we have 10 items in single direction
# shape of a2: (3, 4) means we have 3 rows and 4 columns
# shape of a3: (2, 2, 2) means there are two 2D arrays of shape 2,2
```

```python
# size -> tells how many items
print(f"size of a1: {a1.size}")
print(f"size of a2: {a2.size}")
print(f"size of a3: {a3.size}")
```

```python
# itemsize -> how much space array is occupying in memory and give output in bytes
# a2 a3 both gave the result '8' because integer and float value usually holds 8bytes in memory (in google colab , both int64(by default) and float hold same size)
# if i change it to int32, it will give 4 as output
print(f"itemsize of a1: {a1.itemsize}")
print(f"itemsize of a2: {a2.itemsize}")
print(f"itemsize of a3: {a3.itemsize}")


print("\nNOTE:int32 take 4bytes and int64 take 8bytes")
```

```python
# dtype
print(f"dtype of a1: {a1.dtype}")
print(f"dtype of a2: {a2.dtype}")
print(f"dtype of a3: {a3.dtype}")
```

##ARRAY OPERATIONS

###Vector Operations

##Array Functions

```python
a1 = np.random.random((3,3))
a1 = np.round(a1 * 100)
a1
```

```python
# We will try to cover all the important mathematical functions , we don't need to cover all of them.

# min/max/sum/prod
# min -> give smallest element of array , max -> give largest element , sum sums all the elements , prod multiply all the elements
print(f"max of a1: {np.max(a1)}\n")
print(f"min of a1: {np.min(a1)}\n")
print(f"sum of a1: {np.sum(a1)}\n")
print(f"prod of a1: {np.prod(a1)}\n")
```

```python
# max/min but for each column ( use axis = 0 ) and rows ( use axis = 1)
# this axis thing is applicable to every function
np.max(a1, axis = 0)
```

```python
np.max(a1,axis = 1)
```

```python
# mean/median/std/var
print(f"mean of a1: {np.mean(a1)}\n")
print(f"median of a1: {np.median(a1)}\n")
print(f"std of a1: {np.std(a1)}\n")
print(f"var of a1: {np.var(a1)}\n")
```

```python
# trignometric fucntions ( we dont use it very often )
np.sin(a1)
```

```python
# dot product
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3)

print(f"a2 vector:\n{a2}\n")
print(f"a3 vector:\n{a3}\n")

# Dot product condition satisfied! a2 of shape 3,4 and a3 of shape 4,3
a4 = np.dot(a2,a3)
print(f"dot product of a2 and a3:\n{a4}.\n\nShape of a4 is {np.shape(a4)}")
```

```python
# log and exponents
print(np.log(a1))
print()
print(np.exp(a1))
```
